```python
# تقسيم الاعمده

# الأعمدة النصية (object) >> هذي الأعمدة تحتوي على بيانات نصية مثل اسم الأغنية، اسم الفنانين، أو تاريخ الإصدار.
# نستخدم select_dtypes لتحديد الأعمدة من نوع object
print("الأعمدة النصية (object):")
# select_dtypes: دالة تختار الأعمدة اللي نوعها object (نصوص) من الداتا فريم
# columns.tolist(): تحول أسماء الأعمدة إلى قائمة
print(df2.select_dtypes(include=['object']).columns.tolist())
# بعد طباعة الأعمدة، نطبع خط فاصل بين النتائج
print("="*50)

# الأعمدة العددية الصحيحة (int64) >> هذي الأعمدة تحتوي على أرقام صحيحة بدون كسور (مثل المدة بالمللي ثانية، الشعبية، السلم الموسيقي).
# نستخدم نفس الدالة select_dtypes لكن نختار int64 (أرقام صحيحة).
print("الأعمدة العددية الصحيحة (int64):")
# select_dtypes: نحدد الأعمدة اللي نوعها int64 (أرقام صحيحة)
# columns.tolist(): نعرض أسماء الأعمدة على شكل قائمة
print(df2.select_dtypes(include=['int64']).columns.tolist())
# نفس الشيء، نطبع خط فاصل بين النتائج
print("="*50)

# الأعمدة العددية العشرية (float64) >> هذي الأعمدة تحتوي على أرقام تحتوي على كسور أو فاصلة عشرية (مثل danceability, energy, loudness, tempo).
# نستخدم select_dtypes لاختيار الأعمدة من نوع float64.
print("الأعمدة العددية العشرية (float64):")
# select_dtypes: نحدد الأعمدة اللي نوعها float64 (أرقام تحتوي على كسور)
# columns.tolist(): نعرض أسماء الأعمدة على شكل قائمة
print(df2.select_dtypes(include=['float64']).columns.tolist())

# الفائدة:
# - نقدر نستخدم هذي المعلومات عشان نفهم نوع البيانات اللي نتعامل معها.
# - هذا يساعدنا على اتخاذ قرارات صحيحة في حال أردنا تنظيف البيانات أو تعديلها.
# - مثلًا لو كان فيه عمود من نوع object وكان المفروض يكون من نوع رقمي، نقدر نعدله.
# - لو كنا نحتاج لعملية تحويل أو تحليل البيانات، نكون عارفين نوع الأعمدة عشان نستخدم الأساليب الصحيحة.
```

```python
# نخلي أسماء الأعمدة كلها حروف كبيرة + نشيل المسافات ونستبدلها بشرطة سفلية "_"
# 1. `str.strip()`: يحذف المسافات الفارغة الزائدة في بداية ونهاية اسم العمود
# 2. `str.upper()`: يحول جميع الحروف إلى حروف كبيرة (uppercase) عشان نضمن التناسق
# 3. `str.replace(' ', '_')`: يحول المسافات بين الكلمات إلى شرطة سفلية "_" عشان نقدر نستخدمها بسهولة في الكود
# 4. `str.replace('-', '_')`: يحول العلامة "-" إلى شرطة سفلية "_" في حال كان فيه فاصل بين الكلمات
df2.columns = df2.columns.str.strip().str.upper().str.replace(' ', '_').str.replace('-', '_')

# نعرض أسماء الأعمدة بعد التنظيف عشان نتأكد من التغيير
# الفائدة: نطبع أسماء الأعمدة بعد ما نظفناها، نتأكد إنها أصبحت متناسقة بدون مسافات أو حروف صغيرة
print("أسماء الأعمدة بعد التنظيف:")
print(df2.columns.tolist())
```

```python
# نحول مدة الأغنية من ملي ثانية إلى دقائق
# الفايدة: يسهل التحليل بدل ما نشتغل على أرقام كبيرة بالمللي ثانية (1000 ملي ثانية = 1 ثانية، و60 ثانية = 1 دقيقة)
df2['duration_min'] = df2['DURATION_MS'] / (1000 * 60)
```

```python
# نصنف الشعبية إلى ثلاث مستويات: منخفض، متوسط، عالي
# الفائدة: يخلي التحليل أوضح بدل أرقام من 0 إلى 100
df2['popularity_level'] = pd.cut(
    df2['POPULARITY'],
    bins=[-1, 30, 70, 100],
    labels=['Low', 'Medium', 'High']
)
```

```python
# نستخرج العقد من سنة الإصدار (مثال: 1995 تصير 1990s)
# الفائدة: نقدر نحلل التغيرات عبر العقود بدال السنوات الكثيرة
df2['decade'] = (df2['year'] // 10) * 10
```

```python
# نحسب كم أغنية في كل عقد
print(df2['decade'].value_counts().sort_index())
```

```python
# نرسم عدد الأغاني في كل عقد (bar chart)
# الفايدة: نعرف أي عقد كان فيه أكثر إنتاج موسيقي
df2['decade'].value_counts().sort_index().plot(kind='bar',color='gray', figsize=(10,5))

# عنوان الرسم بالإنقلش
plt.title("Number of Songs per Decade")

# نسمي المحاور
plt.xlabel("Decade")   # العقد
plt.ylabel("Number of Songs")  # عدد الأغاني

# نعرض الشكل
plt.show()
```

```python
# تجميع البيانات حسب الفنان وحساب المتوسط لكل ميزة صوتية
# الفائدة: نعرف البصمة الصوتية أو الطابع الموسيقي لكل فنان

artist_features = df2.groupby('ARTISTS')[[
    # تحديد الميزات الصوتية التي سيتم حساب المتوسط لها
    'DANCEABILITY',   # قابلية الرقص
    'ENERGY',         # الطاقة
    'SPEECHINESS',    # الكلامية (كمية الكلام في الأغنية)
    'ACOUSTICNESS',   # العزل الصوتي (الموسيقى الغير معالجة إلكترونيًا)
    'INSTRUMENTALNESS', # القيمية الآلية (هل تحتوي على أدوات موسيقية فقط)
    'LIVENESS',        # الحيوية (الطابع الحي للأغنية)
    'VALENCE',         # الفرح (مدى سعادة الأغنية)
    'TEMPO'            # الإيقاع (سرعة الأغنية)
]].mean()  # حساب المتوسط لكل ميزة صوتية لكل فنان
```

```python
# الرقص (Danceability): الأغاني ذات الشعبية العالية تظهر قيم قريبة من الصفر،
# بينما الأغاني ذات الشعبية المنخفضة لها قيم أعلى بكثير.

# الطاقة (Energy): هناك اختلاف بسيط بين الأغاني الشعبية العالية والمنخفضة في هذه الميزة.
# الأغاني الشعبية العالية قليلاً أعلى من الأغاني الشعبية المنخفضة في متوسط الطاقة.

# الفرح (Valence): الأغاني الشعبية العالية تظهر قيم مرتفعة قليلاً مقارنة بالأغاني الشعبية المنخفضة.

# الضوضاء (Loudness): هناك فرق كبير في متوسط الضوضاء بين الأغاني ذات الشعبية العالية والمنخفضة،
# حيث تكون الأغاني الشعبية المنخفضة أكثر هدوءًا بكثير مقارنة بالأغاني ذات الشعبية العالية.
```

```python
# هنا نحدد الأعمدة الرقمية اللي نحتاج نحولها (زي DANCEABILITY, ENERGY, VALENCE ... إلخ)
numerical_cols = ['DANCEABILITY', 'ENERGY', 'VALENCE', 'SPEECHINESS',
                  'ACOUSTICNESS', 'INSTRUMENTALNESS', 'LIVENESS', 'TEMPO']

# نحول كل القيم في الأعمدة هذي إلى أرقام (numeric)
# لو واجه نص أو قيمة ما تنفع تتحول، يخليها NaN مؤقت
df2[numerical_cols] = df2[numerical_cols].apply(pd.to_numeric, errors='coerce')

# بعد التحويل، لو طلع فيه قيم ناقصة (NaN) نعوضها بصفر
# عشان ما تسبب مشاكل في التحليل أو الرسوم البيانية
df2[numerical_cols] = df2[numerical_cols].fillna(0)

# بالنسبة لعمود الشعبية (popularity_level):
# نضيف فئة جديدة باسم "Unknown" عشان نستوعب القيم المفقودة
df2['popularity_level'] = df2['popularity_level'].cat.add_categories('Unknown')

# بعدها أي قيمة ناقصة (NaN) تتحول إلى "Unknown"
df2['popularity_level'] = df2['popularity_level'].fillna('Unknown')
```

```python
# بعد ما خلصنا تنظيف البيانات، نحسب مصفوفة الارتباط (correlation matrix)
# الهدف: نعرف وش العلاقة بين كل ميزة رقمية والثانية (قوة واتجاه العلاقة).

# نحدد الأعمدة الرقمية بس (float و int)، لأن الارتباط ينحسب بين الأرقام مو النصوص.
numerical_df = df2.select_dtypes(include=['float64', 'int64'])
correlation_matrix = numerical_df.corr()

# نرسم خريطة حرارية (heatmap) لمصفوفة الارتباط
# اسم الرسم: "Correlation Matrix of Numerical Features"
# الرسم يوضح العلاقات بين الأعمدة الرقمية:
# - 1 يعني علاقة قوية موجبة (كل ما زاد عمود زاد الثاني)
# - -1 يعني علاقة قوية سالبة (كل ما زاد عمود نقص الثاني)
# - 0 يعني ما فيه علاقة واضحة.
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Correlation Matrix of Numerical Features")
plt.show()
```

```python

# - الأحمر الغامق: ارتباط موجب قوي (إذا زاد واحد، يزيد الثاني معه).
# - الأزرق الغامق: ارتباط سالب قوي (إذا زاد واحد، الثاني ينقص).
# - الألوان الفاتحة أو الرمادية: معناها مافيه ارتباط واضح.

# - Popularity مع Year = 0.55: كل ما الأغنية كانت أحدث، غالبًا شعبيتها أعلى (ارتباط موجب/لون أحمر).

# - Popularity مع Acousticness = -0.37: الأغاني الصوتية غالبًا شعبيتها أقل (ارتباط سالب/لون أزرق).

# - Energy مع Loudness = 0.76: الأغاني اللي فيها طاقة عالية عادة تكون صاخبة (ارتباط موجب قوي/لون أحمر).
# - Danceability مع Valence = 0.37: الأغاني اللي تنرقص عليها غالبًا تكون فرِحة أكثر (ارتباط موجب/لون أحمر فاتح).
# - Year مع Decade = 0.99: طبيعي قوي جدًا لأن العقد يعتمد على السنة (ارتباط موجب قوي/لون أحمر غامق).
```

```python
# نحسب ارتباط كل ميزة مع الشعبية (POPULARITY)
# - موجب عالي: كل ما زادت الميزة تزيد الشعبية (مثال: Year ↑ = شعبية ↑)
# - سالب عالي: كل ما زادت الميزة تنقص الشعبية (مثال: Acousticness ↑ = شعبية ↓)
# - قريب من 0: ما له تأثير واضح
correlation_with_popularity = correlation_matrix['POPULARITY'].sort_values(ascending=False)

print("أقوى الارتباطات مع درجة الشعبية:")
print(correlation_with_popularity)
```
